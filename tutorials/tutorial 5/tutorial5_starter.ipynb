## 1. Exploratory Data Analysis

### What is Exploratory Data Analysis?

**Exploratory Data Analysis (EDA)** is the process of investigating a dataset to understand its structure, quality, patterns, and relationships before performing more advanced analysis or modelling.

EDA is essentially **detective work**: we examine the data to understand what is present, what might be missing, and whether the data makes sense in its domain context.

When performing EDA, we should ask questions about:

* **Structure** — What does the dataset look like? What are its rows and columns?
* **Granularity** — What does each observation represent?
* **Scope** — Does the dataset cover the population, geographic region, and time period of interest?
* **Temporality** — How is the data situated in time?
* **Spatiality** — How is the data situated geographically?
* **Faithfulness** — Does the data appear to accurately represent reality?
* **Domain Context** — Are there domain-specific considerations that affect how we interpret the data?

### Why is EDA important?

Before analyzing or modelling a dataset, we need to understand what the data actually represents. EDA can help us identify:

* Missing values
* Duplicate observations
* Unexpected or invalid values
* Outliers
* Incorrect data types
* Inconsistent categories
* Unexpected relationships between variables
* Patterns or trends that may require further investigation

## 3. Introduction to Time Series

### What is Time Series Data?

**Time series data** is data in which observations are associated with specific points or intervals in time.

For example, we might record:

* Number of COVID-19 cases each day
* Temperature every hour
* Stock prices each day
* Electricity demand every 15 minutes
* Traffic volume each hour

A key feature of time series data is that **the order of observations matters**. Unlike a typical cross-sectional dataset, observations are naturally arranged according to time.

### Why is Time Important?

When working with time-dependent data, we need to consider questions such as:

* When was the observation recorded?
* What time zone was used?
* What time interval does each observation represent?
* Are observations recorded hourly, daily, weekly, or monthly?
* Are there trends over time?
* Are there repeating seasonal or periodic patterns?
* How does the quantity change from one period to the next?

Pandas provides specialized tools for working with datetime data, including:

* `DatetimeIndex`
* Date-based filtering
* `.resample()`
* `.dt` datetime properties
* Percentage change calculations

For example, in a COVID-19 dataset, a missing value and zero confirmed cases could have very different meanings depending on how the data was collected.

### Interquartile Range (IQR)

The **Interquartile Range (IQR)** measures the spread of the middle 50% of the data.

It is calculated as:

$$
IQR = Q_3 - Q_1
$$

where:

- $Q_1$ is the first quartile (25th percentile);
- $Q_3$ is the third quartile (75th percentile).

The IQR is represented by the **width of the box** in a box plot.

A larger IQR indicates that the middle 50% of the observations are more spread out, while a smaller IQR indicates that they are more concentrated.

The IQR is also useful because it is less sensitive to extreme values than measures such as the range.

- `.sum()` — total value;
