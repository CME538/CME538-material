# Boolean Array Selection

A very common data science task is **filtering**: keeping only the rows that satisfy some condition.

Pandas allows `[]` and `.loc[]` to use Boolean (`True`/`False`) values as selectors.

Let's first see the basic idea with our 10-row sample.

````python
elections_sample
````

````python
boolean_list = [
    False, False, False, False, True,
    False, False, True, True, False
]

elections_sample.loc[boolean_list]
````

There is one Boolean value for each row.

- `True` means **keep this row**.
- `False` means **do not keep this row**.

The result contains only the rows corresponding to `True`.

In practice, we usually do not type the Boolean values ourselves. Instead, we **create them from a condition**.

---

## Creating a Boolean Series from a condition

Suppose we want to know which rows correspond to election winners.

We can compare the `Result` Series with the string `"win"`:

````python
# TODO: Check which rows have "win" in the "Result" column.
# The comparison returns True or False for each row.

````

The comparison is performed **element by element**.

For every row, Pandas asks:

> Is the value in `Result` equal to `"win"`?

The result is a Boolean Series with the same Index as the original data.

Usually, we write the condition directly inside the brackets:

This pattern is extremely common in Pandas:

```python
dataframe[condition]
```

You can read it as:

> Return the rows of this DataFrame for which the condition is `True`.

---

## Combining multiple conditions

Often one condition is not enough.

Pandas uses:

- `&` for **AND**,
- `|` for **OR**, and
- `~` for **NOT**.

When combining conditions, put each individual comparison inside parentheses.

````python
elections[
    (elections["Result"] == "win") &
    (elections["%"] < 50)
]
````

This keeps rows where **both** conditions are true:

1. the candidate won, **and**
2. the candidate received less than 50% of the vote.

````python
elections[
    (elections["Result"] == "win") &
    ~(elections["%"] < 50)
]
````

The `~` operator negates a Boolean condition. In other words, `True` becomes `False` and `False` becomes `True`.

The `|` operator is used when **either** condition may be true.

````python
elections[
    (elections["Party"] == "Democratic") |
    (elections["Party"] == "Republican")
]
````

### A cleaner way to test several possible values: `.isin()`

When we want to ask whether a value belongs to a collection of possibilities, `.isin()` is often clearer than several equality comparisons joined with `|`.

## `.loc[]` or `.iloc[]`?

We will use both throughout the course.

`.loc[]` is often easier to read because the code explicitly names the labels being selected. For example:

```python
df.loc[:, ["Candidate", "Year"]]
```

tells the reader exactly which columns are wanted.

With `.iloc[]`:

```python
df.iloc[:, [0, 3]]
```

the reader has to know which variables happen to occupy positions `0` and `3`.

Also, if the order of columns changes, position-based code can select something different, while label-based code continues to ask for the same named columns.

Use `.iloc[]` when **position itself** is what matters.
